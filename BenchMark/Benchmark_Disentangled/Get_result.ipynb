# SCIB 计算指标

```python
import scanpy as sc
from sklearn.metrics import silhouette_score
from sklearn.metrics import adjusted_rand_score, normalized_mutual_info_score, davies_bouldin_score
from sklearn.neighbors import kneighbors_graph
import scib
import numpy as np

def neighbor_score(adata, embeddingA, type_label):
    embedding_A = [i.indices for i in embeddingA]
    neighbor_label = adata.obs.iloc[np.concatenate(embedding_A, axis=0)][type_label]
    cell_label = adata.obs.iloc[np.repeat(np.arange(len(embedding_A)), [len(j) for j in embedding_A])][type_label]
    n_shared_labels = (neighbor_label.values == cell_label.values).sum() / len(cell_label)
    return n_shared_labels

def get_label_score(adata, embedding, type_label):
    
    neighbors = kneighbors_graph(embedding, 50, mode='distance', include_self=True, n_jobs=25)
    label_score = neighbor_score(adata, neighbors, type_label)
    return label_score

def run_metrics_overall(adata, cell_type_key, cluster_key, embed_key ,batch_key):
    """
    计算总体指标（不分 stage）
    
    Parameters
    ----------
    adata : AnnData
        单细胞数据对象
    cell_type_key : str
        真实标签在 obs 中的列名
    cluster_key : str
        预测标签在 obs 中的列名
    embed_key : str
        用于评估的 embedding 在 obsm 中的 key




    ari = adjusted_rand_score(adata.obs['name.simple'],adata.obs['UNAGI'] )
    nmi =  normalized_mutual_info_score(adata.obs['name.simple'],adata.obs['UNAGI'])
    dbi = davies_bouldin_score(adata.obsm['z'],adata.obs['leiden'])
    label_score = get_label_score(adata,adata.obsm['z'],'name.simple')
    silhouette = silhouette_score(adata.obsm['z'], adata.obs['leiden'])
    isolated_asw = scib.metrics.isolated_labels_asw(adata,label_key = 'name.simple', batch_key = 'stage', embed='z')
    celltype_asw = silhouette_score(adata.obsm['z'], adata.obs['name.simple'])
    isolated_f1 = scib.metrics.isolated_labels_f1(adata,label_key = 'name.simple', batch_key = 'stage', embed='z')
    clisi_graph = scib.metrics.clisi_graph(adata, type_='embed', label_key='name.simple', batch_key='stage', use_rep='z')
    Returns
    -------
    dict : 各项指标结果
    """
    
    ari = adjusted_rand_score(adata.obs[cell_type_key], adata.obs[cluster_key])
    nmi = normalized_mutual_info_score(adata.obs[cell_type_key], adata.obs[cluster_key])
    dbi = davies_bouldin_score(adata.obsm[embed_key], adata.obs[cluster_key])
    label_score = get_label_score(adata, adata.obsm[embed_key], cell_type_key)
    silhouette = silhouette_score(adata.obsm[embed_key], adata.obs[cluster_key])
    isolated_asw = scib.metrics.isolated_labels_asw(adata, label_key=cell_type_key, batch_key=batch_key, embed=embed_key)
    celltype_asw = silhouette_score(adata.obsm[embed_key], adata.obs[cell_type_key])
    isolated_f1 = scib.metrics.isolated_labels_f1(adata, label_key=cell_type_key, batch_key=batch_key, embed=embed_key)
    
    sc.pp.neighbors(adata, use_rep=embed_key, n_neighbors=50)
    # 计算 cLISI (cell-type LISI)
    clisi = scib.metrics.clisi_graph(adata,label_key=cell_type_key,  type_="embed",use_rep=embed_key)
    
    results = {
        "ARI": ari,
        "NMI": nmi,
        "DBI": dbi,
        "Label score": label_score,
        "Silhouette (cluster)": silhouette,
        "Silhouette (cell type)": celltype_asw,
        "Isolated ASW": isolated_asw,
        "Isolated F1": isolated_f1,
        "Clisi_graph":clisi,
        "Overall SCIB": (ari + nmi + celltype_asw + isolated_f1 + isolated_asw + clisi) / 6
    }
    
    for k,v in results.items():
        print(f"{k}: {v}")
    
    return results



metrics = []
for i in range(len(adatas_lst)):
    adata = adatas_lst[i]
    print("Caculate ",i, " metric")
    metric = run_metrics_overall(adata, cell_type_key='name.simple', cluster_key='pred', embed_key='embedding',batch_key='stage')
    metrics.append(metric)
import pandas as pd
# 方法名称
methods = ['scGNN','scGGAN','GraphSCC','scVI','UNAGI','Caunagi','scGEN','scGPT','Geneformer']
# 转换为 DataFrame
df = pd.DataFrame(metrics, index=methods).T
df.to_csv("/disk1/cai029/Caufussion/Benchmark_Disentangled/Metrics_result/result_metrics_10.csv")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from pandas.plotting import parallel_coordinates

# 假设 df 已经存在

# 1. 每个方法在每个指标取平均值
df_mean = df.groupby(["methods", "metrics"])["score"].mean().reset_index()

# 2. pivot 到宽表
df_pivot = df_mean.pivot(index="methods", columns="metrics", values="score").reset_index()

# 3. 平行坐标图需要“长表”，所以 melt 一下
df_long = pd.melt(df_pivot, id_vars=["methods"], var_name="metrics", value_name="score")

# 4. 画图
plt.figure(figsize=(12, 6))
parallel_coordinates(df_pivot, class_column="methods", cols=df_pivot.columns[1:], alpha=0.7)

plt.xticks(rotation=30, fontsize=12)
plt.ylabel("Score", fontsize=12)
plt.title("Parallel Coordinates Plot of Methods vs Metrics", fontsize=14)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
