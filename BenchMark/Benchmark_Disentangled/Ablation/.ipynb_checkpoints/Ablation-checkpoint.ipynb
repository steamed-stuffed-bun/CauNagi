```python
SEEDS = [888, 999, 101, 202, 303]
```

```python
import sys
sys.path.append('/disk1/cai029/CauNagi/Benchmark/Benchmark_Disentangled/Ablation/Caunagi/')
from Caunagi import caunagi_main
import os
os.environ["JAVA_TOOL_OPTIONS"] = "-Djava.awt.headless=true"
```

```python
concept_list = ['stage','CellType']
concept_cdag = [[0,0,0],[1,0,0],[0,0,0]]
state_idx = {0:0,1:1,2:2,3:3}

model = caunagi_main.Caunagi(
                    concept_list,
                    concept_cdag,
                    total_stage_num=4,
                    save_and_sample_every=1000)
iDREM_Path = '/disk1/cai029/biosoft/idrem'
```

```python
import os
iter_num = 2
for iteration in range(iter_num):
    
    if iteration == 0:
        train_data_path = '/disk1/cai029/CauNagi/Benchmark/Benchmark_Disentangled/Ablation/use_data/'
    else:
        train_data_path = '/disk1/cai029/CauNagi/Benchmark/Benchmark_Disentangled/Ablation/use_data/temp_path'
    print("iteration : " ,iteration)
    model.process_data(data_path= train_data_path,
                      temp_path = '/disk1/cai029/CauNagi/Benchmark/Benchmark_Disentangled/Ablation/use_data/temp_path',
                      stage_name='stage',
                      iteration=iteration,
                      celltype_concept_name = 'CellType',
                      disease_idx=state_idx
                     )
    model.setup_train(model_save_path="/disk1/cai029/CauNagi/Benchmark/Benchmark_Disentangled/Ablation/use_data/model_save/", 
                        seed = SEEDS[4],
                        epoch_num=3000,max_profile_size=3000)
   
    model.run_caunagi(iDREM_Path)
```
