```python
import pandas as pd
import os
for i in range(10):
    if os.path.isfile("./WMDS_output/"+str(i)+"_WMDS_net_gene_list.csv"):
        driver_lst = pd.read_csv("./WMDS_output/"+str(i)+"_WMDS_net_gene_list.csv")
    if os.path.isfile("./WMDS_output/"+str(i)+"_WMDS_net_gene_scores.csv"):  
        dirver_score = pd.read_csv("./WMDS_output/"+str(i)+"_WMDS_net_gene_scores.csv")
    result = pd.concat([driver_lst, dirver_score], axis=1)
    result.to_csv("./WMDS_output/"+str(i)+"_WMDS_net_result.csv")
```
