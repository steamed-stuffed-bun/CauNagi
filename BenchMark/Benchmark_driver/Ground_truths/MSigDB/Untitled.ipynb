```python
import pandas as pd

# 函数：加载GMT
def load_gmt(file_path):
    data = []
    with open(file_path, 'r') as f:
        for line in f:
            parts = line.strip().split('\t')
            set_name = parts[0]
            genes = parts[2:]
            for gene in genes:
                data.append({'set_name': set_name, 'gene': gene})
    return pd.DataFrame(data)

# 示例：加载并过滤
keywords = ['aml', 'leukemia', 'myeloid', 'transcription', 'regulation', 'runx1', 'cebpa']
df = load_gmt('c3.tft.v2023.1.Hs.symbols.gmt')  # 替换你的文件
aml_df = df[df['set_name'].str.lower().str.contains('|'.join(keywords))]

# 提取独特基因
unique_genes = aml_df['gene'].unique()
unique_df = pd.DataFrame({'gene_name': unique_genes})

# 保存
unique_df.to_csv('msigdb_aml_genes.csv', index=False)
print(unique_df.head(10))  # 检查
```
