```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from upsetplot import UpSet
from upsetplot import from_memberships
import gseapy as gp
from scipy import stats
import warnings
warnings.filterwarnings('ignore')



def process_dataframe(df):
    # 1. 假设你的原始数据名为 df
    # 如果你的 index 已经是 0, 1, 2... (代表 rank)
    # 如果没有，可以先 df.reset_index(drop=True) 确保索引正确
    
    # 2. 将行索引(rank)转换为一列，方便后续处理
    df_with_rank = df.reset_index().rename(columns={'index': 'rank'})
    
    # 3. 使用 melt 将多列转换为行
    df_long = df_with_rank.melt(
        id_vars='rank', 
        var_name='run_id', 
        value_name='gene'
    )
    
    # 4. 调整数值，使其从 1 开始 (1-based)
    df_long['rank'] = df_long['rank'] + 1
    df_long['run_id'] = df_long['run_id'].astype(int) + 1
    
    # 5. 按照你的要求整理列顺序
    df_final = df_long[['run_id', 'gene', 'rank']]
    
    # 6. (可选) 按 run_id 和 rank 排序，让数据更整齐
    df_final = df_final.sort_values(['run_id', 'rank']).reset_index(drop=True)
    
    print(df_final)
    return df_final

# 设置绘图风格
sns.set_style('whitegrid')
plt.rcParams['figure.dpi'] = 150

# 读取 Ground Truth
ground_truth = set(driver_lst)

# 定义方法名称和对应文件
methods = {
    'Caunagi': caunagi_df,
    'CellOracle': celloracle_df,
    'WMDS.net': WMDS_df,
    'Caufinder': caufinder_df,
    'CEFCON': cefcon_df
}

# 读取所有结果，存储为字典：{method: {run_id: [(gene, rank), ...]}}
all_results = {}
for method, file in methods.items():
    print("============process Method ",method,"  ============")
    df = process_dataframe(file)
    # 确保按 run_id 和 rank 排序
    df = df.sort_values(['run_id', 'rank']).reset_index(drop=True)
    # 构建嵌套字典
    runs = {}
    for run_id, group in df.groupby('run_id'):
        genes = group['gene'].tolist()
        # 如果工具未提供 rank，则默认顺序即为 rank
        runs[run_id] = genes
    all_results[method] = runs

print("数据加载完成。")
```

```python
def compute_metrics(genes, ground_truth, top_k=None):
    """
    计算精确率、召回率、F1。
    如果 top_k 不为 None，则只考虑前 top_k 个基因。
    """
    if top_k is not None:
        genes = genes[:top_k]
    genes_set = set(genes)
    tp = len(genes_set & ground_truth)
    fp = len(genes_set) - tp
    fn = len(ground_truth) - tp
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0
    return precision, recall, f1

# 收集所有指标
records = []
for method, runs in all_results.items():
    for run_id, genes in runs.items():
        # 使用全部基因计算（您也可以限制 top N，例如 top 200）
        p, r, f = compute_metrics(genes, ground_truth)
        records.append({
            'Method': method,
            'Run': run_id,
            'Precision': p,
            'Recall': r,
            'F1': f
        })

df_metrics = pd.DataFrame(records)

# 计算均值和标准差用于条形图
df_summary = df_metrics.groupby('Method')[['Precision', 'Recall', 'F1']].agg(['mean', 'std']).reset_index()
# 扁平化列名
df_summary.columns = ['Method', 'Precision_mean', 'Precision_std', 'Recall_mean', 'Recall_std', 'F1_mean', 'F1_std']

# 绘制条形图（分组条形图）
# 绘制条形图
fig, ax1 = plt.subplots(figsize=(4, 6))

x = np.arange(len(df_summary['Method']))
width = 0.25

ax1.bar(x - width, df_summary['Precision_mean'], width,
        yerr=df_summary['Precision_std'], capsize=3, label='Precision')

ax1.bar(x, df_summary['Recall_mean'], width,
        yerr=df_summary['Recall_std'], capsize=3, label='Recall')

ax1.bar(x + width, df_summary['F1_mean'], width,
        yerr=df_summary['F1_std'], capsize=3, label='F1')

ax1.set_xticks(x)
ax1.set_xticklabels(df_summary['Method'], rotation=45, ha='right')
ax1.set_ylabel('Score')
ax1.set_title('Performance Metrics (Mean ± SD)')
ax1.legend()
ax1.set_ylim(0, 1)

plt.tight_layout()
plt.savefig("Precision_Recall_F1.png", dpi=300, bbox_inches='tight')
plt.savefig("Precision_Recall_F1.pdf", dpi=400, bbox_inches='tight')
plt.show()
```

```python
# 构建每个方法的高置信度基因集
consensus_sets = {}
for method, runs in all_results.items():
    # 统计每个基因在10次运行中出现的次数
    gene_counts = {}
    for genes in runs.values():
        for g in genes:
            gene_counts[g] = gene_counts.get(g, 0) + 1
    # 选择出现次数 >=5 的基因
    consensus = {g for g, cnt in gene_counts.items() if cnt >= 5}
    consensus_sets[method] = consensus

# 添加 Ground Truth 作为一个集合
consensus_sets['Ground Truth'] = set(driver_lst)

# 转换为 upsetplot 所需的格式：每个样本属于哪些集合
# 构建所有基因的列表
all_genes = set()
for s in consensus_sets.values():
    all_genes.update(s)

# 为每个基因记录它所属的集合
memberships = []
for gene in all_genes:
    sets = [name for name, genes in consensus_sets.items() if gene in genes]
    memberships.append(sets)

# 创建 multi-index 数据框
data = from_memberships(memberships, data=pd.Series(1, index=range(len(memberships))))

# 绘制 UpSet 图
upset = UpSet(data, subset_size='count', show_counts='%d', sort_by='cardinality')
upset.plot()
plt.savefig("Method concordance.png", dpi=300, bbox_inches='tight')
plt.savefig("Method concordance.pdf", dpi=400, bbox_inches='tight')
plt.show()
```

```python
import gseapy as gp
import pandas as pd

# 下载库
gene_sets = gp.get_library(name="KEGG_2021_Human")

# 你要找的 7 个 pathway（用小写、空格形式作为匹配模式）
wanted_patterns = {
    "AML": "acute myeloid leukemia",
    "Hematopoietic_lineage": "hematopoietic cell lineage",
    "Transcriptional_misregulation": "transcriptional misregulation in cancer",
    "MAPK": "mapk signaling pathway",
    "PI3K_AKT": "pi3k-akt signaling pathway",   # 注意 KEGG 名里可能有 '-' 或 空格
    "JAK_STAT": "jak-stat signaling pathway",
    "Cell_cycle": "cell cycle"
}

# 辅助函数：在 gene_sets 的 keys 中找到最合适的 key（容错：大小写、下划线、前缀 KEGG_ 等）
def find_key_for_pattern(gs_keys, pattern):
    pattern_l = pattern.lower()
    # 1) 直接包含（最常见：'Acute myeloid leukemia'）
    exact_matches = [k for k in gs_keys if pattern_l in k.lower()]
    if len(exact_matches) == 1:
        return exact_matches[0]
    if len(exact_matches) > 1:
        return exact_matches[0]   # 有多个时返回第一个（通常不会发生）
    # 2) 尝试匹配带下划线 / 带 KEGG_ 前缀 的格式（如 KEGG_ACUTE_MYLOID_LEUKEMIA）
    alt = pattern_l.replace(" ", "_").replace("-", "_").upper()
    alt_keys = [k for k in gs_keys if alt in k.upper()]
    if len(alt_keys) >= 1:
        return alt_keys[0]
    # 3) 没找到
    return None

# 寻找并提取
found = {}
for shortname, pat in wanted_patterns.items():
    key = find_key_for_pattern(list(gene_sets.keys()), pat)
    if key is None:
        print(f"[WARN] 未找到匹配的 pathway for pattern '{pat}'")
    else:
        print(f"[FOUND] {shortname} -> '{key}' ({len(gene_sets[key])} genes)")
        found[shortname] = gene_sets[key]

# 示例：打印 AML 前 10 个基因
if "AML" in found:
    print("AML genes (example):", found["AML"][:10])

```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.path import Path
import numpy as np

# ----------------------------
# 全局样式（发表级别）
# ----------------------------
plt.rcParams['font.sans-serif'] = ['Arial', 'Helvetica', 'DejaVu Sans']
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['axes.linewidth'] = 1.0
plt.rcParams['axes.titlesize'] = 16
plt.rcParams['axes.labelsize'] = 14
plt.rcParams['xtick.labelsize'] = 11
plt.rcParams['ytick.labelsize'] = 11

def get_index(list_a, list_b):
    set_b = set(list_b)
    positions = []
    for index, element in enumerate(list_a):
        if element in set_b:
            positions.append(index)
    return positions

# ----------------------------
# 数据
# ----------------------------
genes = ["STAT3", "GATA2", "STAT1", "SPI1", "RUNX1", "RBBP5", "TP53", "ZNF143", "GATA3", "ZBTB7A",
         "ZEB1", "STAT4", "CHD7", "ATF3", "RXRA", "JAK2", "CEBPA", "EOMES", "PRDM1", "ATF1"]
scores = [7.106, 6.999, 6.897, 6.743, 6.721, 6.625, 6.580, 6.435, 6.412, 6.233,
          5.915, 5.756, 5.726, 5.719, 5.609, 5.102, 4.868, 4.786, 4.775, 4.769]

# 通路到基因索引
pathways = {
    "AML Key Regulators": get_index(genes, found['AML']),
    "PI3K-Akt signaling pathway": get_index(genes, found['PI3K_AKT']),
    "JAK-STAT Signaling": get_index(genes, found['JAK_STAT']),
    "MAPK signaling pathway": get_index(genes, found['MAPK']),
    "Transcriptional Dysregulation": get_index(genes, found['Transcriptional_misregulation'])
}

# 主题色（与示例图相似）
colors = ['#4A7BB7', '#E78736', '#98C25D', '#E25B58', '#C97DA7']

# ----------------------------
# 画布与子图布局
# ----------------------------
fig = plt.figure(figsize=(12, 9))
gs = fig.add_gridspec(2, 1, height_ratios=[1, 1.3], hspace=0.06)
ax1 = fig.add_subplot(gs[0])
ax2 = fig.add_subplot(gs[1])

# ------------------------------------------
# 上图：柱状图（更整洁的配色、zorder、刻度）
# ------------------------------------------
n = len(genes)
x = np.arange(n)
# 生成渐变色（冷到暖）
bar_colors = [plt.cm.viridis(0.15 + i*0.035) for i in range(n)]
bars = ax1.bar(x, scores, color=bar_colors, width=0.6, zorder=5, edgecolor='none')

ax1.set_ylabel('Influence score', fontsize=14, fontweight='bold', labelpad=8)
ax1.set_ylim(0, 8)
ax1.set_xlim(-0.5, n - 0.5)

# x ticks + rotate 标签
ax1.set_xticks(x)
ax1.set_xticklabels(genes, rotation=45, ha='right', fontsize=11)
ax1.set_xticklabels(genes, rotation=90, ha='center', fontsize=11)
# 网格线和边框
ax1.grid(True, linestyle='--', color='#DCDCDC', alpha=0.7, zorder=0)
for spine in ax1.spines.values():
    spine.set_linewidth(1.0)
    spine.set_color('black')

# 在柱子顶部添加数值（可选，发表图常用）
for rect, val in zip(bars, scores):
    height = rect.get_height()
    ax1.text(rect.get_x() + rect.get_width() / 2, height + 0.12,
             f"{val:.2f}", ha='center', va='bottom', fontsize=9)

# ------------------------------------------
# 下图：Sankey-like connections（使用 Path + PathPatch）
# ------------------------------------------
ax2.set_xlim(-0.5, n - 0.5)
ax2.set_ylim(0, 10)
ax2.axis('off')

# 基因下方的小方块（作为连线起点标识）
gene_box_y = 7.95
for i in range(n):
    ax2.add_patch(patches.Rectangle((i - 0.2, gene_box_y), 0.4, 0.20,
                                   color='#F2B134', alpha=0.95, zorder=4))
# 通路方块位置（铺在基因区下方，居中）
pathway_x_pos = np.linspace(1.2, n - 2.2, len(pathways))
pathway_box_y = 2.5
pathway_box_h = 0.6
pathway_box_w = 1.2

for i, (name, linked_genes) in enumerate(pathways.items()):
    px = pathway_x_pos[i]
    color = colors[i % len(colors)]

    # 通路方块
    ax2.add_patch(patches.Rectangle((px - pathway_box_w / 2, pathway_box_y),
                                   pathway_box_w, pathway_box_h,
                                   color=color, alpha=0.95, zorder=3, ec='none'))
    # 通路名（居中，略微旋转）
    ax2.text(px, pathway_box_y - 0.25, name, ha='center', va='top',
             rotation=30, fontsize=12, fontweight='bold')

    # 绘制从每个基因到通路的平滑曲线（cubic Bezier）
    for gene_idx in linked_genes:
        gx = x[gene_idx]

        verts = [
            (gx, gene_box_y),        # start
            (gx, 6.2),               # control point 1 (垂直向下拉)
            (px, 4.0),               # control point 2 (向通路靠拢)
            (px, pathway_box_y + pathway_box_h)  # end (通路顶部)
        ]
        codes = [Path.MOVETO, Path.CURVE4, Path.CURVE4, Path.CURVE4]
        path = Path(verts, codes)

        # 线宽按 score 比例缩放（可读但不过分粗）
        lw = max(0.6, (scores[gene_idx] / max(scores)) * 3.0)
        patch = patches.PathPatch(path, facecolor='none', edgecolor=color,
                                  lw=lw, alpha=0.55, zorder=2, capstyle='round')
        ax2.add_patch(patch)

# ------------------------------------------
# 美化与保存
# ------------------------------------------
plt.tight_layout()
# 保存为高分辨率（示例）
plt.savefig("publication_style_figure.png", dpi=300, bbox_inches='tight')
plt.savefig('./publication_style_figure.pdf', dpi=400, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.path import Path
import numpy as np

# ----------------------------
# 1. 基础配置
# ----------------------------
plt.rcParams['font.sans-serif'] = ['DejaVu Sans', 'Arial']
plt.rcParams['axes.unicode_minus'] = False

# ----------------------------
# 2. 数据准备
# ----------------------------
genes_raw = ["STAT3","GATA2","STAT1","SPI1","RUNX1","RBBP5","TP53","ZNF143","GATA3","ZBTB7A",
             "ZEB1","STAT4","CHD7","ATF3","RXRA","JAK2","CEBPA","EOMES","PRDM1","ATF1"]
scores_raw = [7.106, 6.999, 6.897, 6.743, 6.721, 6.625, 6.580, 6.435, 6.412, 6.233,
              5.915, 5.756, 5.726, 5.719, 5.609, 5.102, 4.868, 4.786, 4.775, 4.769]

sorted_indices = np.argsort(scores_raw)[::-1]
genes = [genes_raw[i] for i in sorted_indices]
scores = [scores_raw[i] for i in sorted_indices]

pathways = {
    "AML Key Regulators": get_index(genes, found['AML']),
    "PI3K-Akt signaling pathway": get_index(genes, found['PI3K_AKT']),
    "JAK-STAT Signaling": get_index(genes, found['JAK_STAT']),
    "MAPK signaling pathway": get_index(genes, found['MAPK']),
    "Transcriptional Dysregulation": get_index(genes, found['Transcriptional_misregulation'])
}
p_colors = ['#4A7BB7', '#E78736', '#98C25D', '#E25B58', '#C97DA7']

# ----------------------------
# 3. 画布布局优化 (局部保留轴线)
# ----------------------------
fig, ax = plt.subplots(figsize=(10, 12))
ax.set_xlim(-4, 16)
ax.set_ylim(len(genes) - 0.5, -1.5)

# 隐藏不需要的边框
for spine in ['left', 'right', 'bottom']:
    ax.spines[spine].set_visible(False)

# --- 💡 关键修改：局部保留顶部轴线 ---
ax.spines['top'].set_visible(True)
ax.spines['top'].set_color('#BBBBBB')
ax.spines['top'].set_linewidth(0.8)
# 只在 0 到 8 之间绘制轴线 (条形图区间)
ax.spines['top'].set_bounds(0, 8) 

ax.xaxis.tick_top()
ax.set_xticks(np.arange(0, 9, 2)) # 仅在条形图上方显示刻度
ax.tick_params(axis='x', colors='#666666', labelsize=10, length=4)

# 移除 Y 轴刻度
ax.set_yticks([])

# 轴标题放在轴线正上方
ax.text(4, -2.2, "Influence Score", ha='center', fontsize=14, fontweight='bold', color='#444444')

# ----------------------------
# 4. 绘制条形图与锚点 (保持优雅)
# ----------------------------
y_pos = np.arange(len(genes))
bar_colors = plt.cm.viridis_r(np.linspace(0.1, 0.7, len(genes)))
ax.barh(y_pos, scores, color=bar_colors, height=0.6, edgecolor='white', linewidth=0.5, zorder=2)

anchor_x_offset, anchor_w = 0.3, 0.25
for i, s in enumerate(scores):
    rect = patches.FancyBboxPatch((s + anchor_x_offset, i - 0.2), anchor_w, 0.4, 
                                  boxstyle="round,pad=0.02", color='#F2B134', zorder=4)
    ax.add_patch(rect)
    ax.text(-0.5, i, genes[i], ha='right', va='center', fontsize=13, fontweight='bold', color='#333333')
    ax.text(s - 0.1, i, f"{s:.2f}", ha='right', va='center', fontsize=10, color='white', fontweight='bold')

# ----------------------------
# 5. 绘制通路方块
# ----------------------------
path_x_start = 12.0
pathway_names = list(pathways.keys())
p_y_pos = np.linspace(1, len(genes)-2, len(pathway_names))
p_box_h, p_box_w = 0.8, 1.5 

p_centers = {}
for i, name in enumerate(pathway_names):
    py = p_y_pos[i]
    color = p_colors[i]
    box = patches.FancyBboxPatch((path_x_start, py - p_box_h/2), p_box_w, p_box_h,
                                 boxstyle="round,pad=0.1", fc=color, ec='white', alpha=0.9, zorder=3)
    ax.add_patch(box)
    ax.text(path_x_start + p_box_w + 0.3, py, name, va='center', fontsize=11, fontweight='bold', color='#444444')
    p_centers[name] = (path_x_start, py)

# ----------------------------
# 6. 绘制贝塞尔连线
# ----------------------------
for i, name in enumerate(pathway_names):
    target_x, target_y = p_centers[name]
    color = p_colors[i]
    for g_idx in pathways[name]:
        start_x = scores[g_idx] + anchor_x_offset + anchor_w
        start_y = g_idx
        dist = (target_x - start_x) * 0.4 # 稍微收紧曲线，让汇聚更平滑
        verts = [(start_x, start_y), (start_x + dist, start_y), (target_x - dist, target_y), (target_x, target_y)]
        path = Path(verts, [Path.MOVETO, Path.CURVE4, Path.CURVE4, Path.CURVE4])
        lw = 0.5 + (scores[g_idx] / max(scores)) * 2.5
        patch = patches.PathPatch(path, facecolor='none', edgecolor=color, lw=lw, alpha=0.35, zorder=1)
        ax.add_patch(patch)

# ----------------------------
# 7. 导出
# ----------------------------
plt.savefig("Sankey_Partial_Axis.png", dpi=300, bbox_inches='tight')
plt.savefig("Sankey_Partial_Axis.pdf", dpi=400, bbox_inches='tight')
plt.show()
```
