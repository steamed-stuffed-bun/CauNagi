```python
import pandas as pd
import scanpy as sc
import anndata as ad
import os
import glob

# 获取所有.dem文件
dem_files = glob.glob("./AML_scRNA-seq/*.dem.txt")
```

```python
dem_files
```

```python
for dem_file in dem_files:
    # 获取对应的anno文件
    base_name = dem_file.replace(".dem.txt", "")
    anno_file = f"{base_name}.anno.txt"
    # print(anno_file)
    if not os.path.exists(anno_file):
        print(f"Warning: No matching annotation file for {dem_file}")
        continue
    
    print(f"Processing {dem_file} and {anno_file}...")
    
    # 读取表达数据
    expr_df = pd.read_csv(dem_file, sep='\t', index_col=0).T
    
    # 读取注释数据
    anno_df = pd.read_csv(anno_file, sep='\t', index_col=0)
    
    # 确保细胞匹配
    common_cells = expr_df.index.intersection(anno_df.index)
    expr_df = expr_df.loc[common_cells]
    anno_df = anno_df.loc[common_cells]
    
    # 创建AnnData对象
    adata = ad.AnnData(X=expr_df.values, var=pd.DataFrame(index=expr_df.columns), obs=anno_df)
    adata.var_names = expr_df.columns
    adata.obs_names = expr_df.index
    
    # 保存
    output_file = f"{base_name}.h5ad"
    adata.write(output_file)
    print(f"Saved to {output_file}")

print("All files processed!")
```
